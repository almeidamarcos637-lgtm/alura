```python
import subprocess, time
from google.colab.output import eval_js
from IPython.display import display, HTML

DISPLAY = ":1"
VNC_PORT = 5900
WEB_PORT = 6080

def run(cmd):
    return subprocess.run(cmd, shell=True, capture_output=True, text=True)

# ============================================
# 1. LIMPEZA
# ============================================
run("pkill -9 -f x11vnc; pkill -9 -f websockify; pkill -9 -f cloudflared; sleep 2")

# ============================================
# 2. INSTALA SE PRECISO
# ============================================
def bin_ok(n):
    return run(f"which {n} >/dev/null 2>&1").returncode == 0

if not (bin_ok("x11vnc") and bin_ok("websockify")):
    print("📦 Instalando x11vnc + noVNC + websockify...")
    run("DEBIAN_FRONTEND=noninteractive apt-get install -y -qq x11vnc novnc websockify")
    print("✅ Instalado.")

# ============================================
# 3. X11VNC (VNC server na tela :1)
# ============================================
print("📡 Subindo x11vnc...")
run(f"nohup x11vnc -display {DISPLAY} -forever -shared -nopw "
    f"-listen 127.0.0.1 -rfbport {VNC_PORT} > /tmp/x11vnc.log 2>&1 &")
time.sleep(3)

r = run(f"ss -ltn | grep :{VNC_PORT}")
if not r.stdout.strip():
    print("❌ x11vnc não subiu. Log:")
    print(run("tail -20 /tmp/x11vnc.log").stdout)
    raise SystemExit
print(f"✅ x11vnc em :{VNC_PORT}")

# ============================================
# 4. WEBSOCKIFY + NOVNC (converte VNC em WebSocket)
# ============================================
print("🌐 Subindo websockify + noVNC...")
run(f"nohup websockify --web /usr/share/novnc/ {WEB_PORT} "
    f"localhost:{VNC_PORT} > /tmp/websockify.log 2>&1 &")
time.sleep(3)

r = run(f"ss -ltn | grep :{WEB_PORT}")
if not r.stdout.strip():
    print("❌ websockify não subiu. Log:")
    print(run("tail -20 /tmp/websockify.log").stdout)
    raise SystemExit
print(f"✅ websockify em :{WEB_PORT}")

# ============================================
# 5. GERA URL PÚBLICA VIA PROXY NATIVO DO COLAB
# ============================================
# 🔑 proxyPort não usa DNS externo — é o próprio Colab roteando
url_proxy = eval_js(f"google.colab.kernel.proxyPort({WEB_PORT})")
link_vnc = f"{url_proxy}/vnc.html?autoconnect=1&resize=scale"

# ============================================
# 6. MOSTRA O CARD COM O LINK
# ============================================
display(HTML(f"""
<div style="
    background:#0f172a;color:#e2e8f0;padding:24px;
    border-radius:12px;font-family:system-ui,sans-serif;
    margin:10px 0;">
  <h2 style="margin:0 0 8px;color:#4ade80;">🖥️ Desktop Linux ao vivo</h2>
  <p style="margin:0 0 16px;color:#94a3b8;">
    Clique no botão. O link é servido pelo próprio Colab — não depende de cloudflared.
  </p>
  <a href="{link_vnc}" target="_blank" style="
      display:inline-block;background:#4ade80;color:#0f172a;
      padding:14px 28px;border-radius:8px;text-decoration:none;
      font-weight:bold;font-size:16px;">
    ▶ Abrir Desktop Remoto
  </a>
  <p style="margin:16px 0 0;font-size:12px;color:#64748b;word-break:break-all;">
    <b>URL:</b> {url_proxy}
  </p>
</div>
"""))

print()
print("=" * 55)
print("✅ TUDO PRONTO")
print("=" * 55)
print(f"🌐 Link: {link_vnc}")
print()
print("💡 Abre em QUALQUER aba do navegador (desktop e celular).")
print("💡 Mouse e teclado nativos — sem botões.")
print("💡 NÃO depende de cloudflared. Funciona de verdade.")
```

```python
import subprocess, time, os

def run(cmd, timeout=300):
    return subprocess.run(cmd, shell=True, capture_output=True, text=True, timeout=timeout)

def bin_ok(n):
    return run(f"which {n} >/dev/null 2>&1").returncode == 0

AUDIO_PORT = 6083

print("🔊 Configurando áudio via cloudflared (sem proxy do Colab)...")
print()

# 1. Garantir que cloudflared existe
if not bin_ok("cloudflared"):
    print("→ Baixando cloudflared...")
    run("wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64 -O /usr/local/bin/cloudflared")
    run("chmod +x /usr/local/bin/cloudflared")
    print("  ✅ cloudflared instalado.")

# 2. Verificar se PulseAudio está rodando
r = run("pactl info 2>&1")
if "Server Name" not in r.stdout:
    print("→ Subindo PulseAudio...")
    run("mkdir -p /run/user/0 && chmod 700 /run/user/0")
    os.environ["XDG_RUNTIME_DIR"] = "/run/user/0"
    run("nohup pulseaudio --daemonize=no --exit-idle-time=-1 "
        "--disallow-exit > /tmp/pulseaudio.log 2>&1 &")
    time.sleep(3)
    run("pactl load-module module-null-sink sink_name=virtual_speaker 2>&1")
    run("pactl set-default-sink virtual_speaker 2>&1")
    print("  ✅ PulseAudio + alto-falante virtual prontos.")
else:
    print("✅ PulseAudio já rodando.")

# 3. Garantir que o sink virtual existe
r = run("pactl list short sinks 2>&1")
if "virtual_speaker" not in r.stdout:
    run("pactl load-module module-null-sink sink_name=virtual_speaker 2>&1")
    run("pactl set-default-sink virtual_speaker 2>&1")
    print("→ Alto-falante virtual criado.")
else:
    print("✅ Alto-falante virtual já existe.")

# 4. Reiniciar Firefox com áudio
print("→ Reiniciando Firefox com áudio...")
run("pkill -9 -f firefox; sleep 2")
run("DISPLAY=:1 PULSE_SERVER=unix:/run/user/0/pulse/native "
    "nohup firefox --no-sandbox -width 1200 -height 700 about:blank "
    "> /tmp/ff.log 2>&1 &")
time.sleep(8)
print("  ✅ Firefox com áudio.")

# 5. Matar ffmpeg antigo e subir novo
run(f"pkill -9 -f 'ffmpeg.*{AUDIO_PORT}'; sleep 1")

print("→ Iniciando transmissão de áudio via ffmpeg...")
run(f"nohup ffmpeg -f pulse -i virtual_speaker.monitor "
    f"-c:a libopus -b:a 64k -f webm -listen 1 "
    f"http://0.0.0.0:{AUDIO_PORT} > /tmp/ffmpeg_audio.log 2>&1 &")
time.sleep(3)

r = run(f"ss -ltn | grep :{AUDIO_PORT}")
if r.stdout.strip():
    print(f"  ✅ Áudio transmitindo em :{AUDIO_PORT}.")
else:
    print("  ⚠️ ffmpeg não subiu. Log:")
    print(run("tail -20 /tmp/ffmpeg_audio.log").stdout)

# 6. Subir cloudflared apontando pro áudio
print("→ Abrindo túnel cloudflared pro áudio...")
run(f"pkill -9 -f 'cloudflared.*{AUDIO_PORT}'; sleep 1")
run(f"nohup cloudflared tunnel --url http://localhost:{AUDIO_PORT} "
    f"> /tmp/tunnel_audio.log 2>&1 &")

url_audio = None
for i in range(40):
    time.sleep(1)
    r = run("grep -o 'https://[a-z0-9-]*\\.trycloudflare\\.com' /tmp/tunnel_audio.log | head -1")
    if r.stdout.strip():
        url_audio = r.stdout.strip()
        break

if not url_audio:
    print("→ cloudflared não gerou URL. Tentando HTTP/2...")
    run("pkill -9 -f cloudflared; sleep 2")
    run(f"nohup cloudflared tunnel --protocol http2 --url http://localhost:{AUDIO_PORT} "
        f"> /tmp/tunnel_audio.log 2>&1 &")
    for i in range(40):
        time.sleep(1)
        r = run("grep -o 'https://[a-z0-9-]*\\.trycloudflare\\.com' /tmp/tunnel_audio.log | head -1")
        if r.stdout.strip():
            url_audio = r.stdout.strip()
            break

if not url_audio:
    print("❌ cloudflared não gerou URL. Log:")
    print(run("tail -30 /tmp/tunnel_audio.log").stdout)
    raise SystemExit

link_audio = f"{url_audio}/audio.webm"

print()
print("=" * 60)
print("✅ ÁUDIO PRONTO (via cloudflared)")
print("=" * 60)
print()
print(f"🔊 Link de áudio: {link_audio}")
print()
print("💡 Abra o link numa aba do navegador e deixe tocando.")
print("💡 O VNC continua funcionando normalmente na outra aba.")
print()
```
